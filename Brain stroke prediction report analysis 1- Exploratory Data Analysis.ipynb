```python
# BRAIN STROKE DATASET
# EXPLORATORY DATA ANALYSIS (EDA)
# WITH DATA PREPROCESSING

# STEP 1 - IMPORT REQUIRED LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
```

```python
# STEP 2 - SET VISUALIZATION STYLE

sns.set_style("whitegrid")

# Professional color palette
blue_color = "#1F77B4"
green_color = "#2E8B57"
red_color = "#C0392B"
orange_color = "#D68910"
```

```python
# STEP 3 - LOAD THE DATASET

df = pd.read_csv("C:/Users/VIHANGI/Desktop/4th YEAR 1st SEM/brain_stroke.csv")
```

```python
# STEP 4 - INITIAL DATA UNDERSTANDING

print("First 5 Rows")
print(df.head())

print("\nDataset Shape")
print(df.shape)

print("\nColumn Names")
print(df.columns)

print("\nData Types")
print(df.dtypes)

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Values")
print(df.duplicated().sum())
```

```python
# STEP 5 - REMOVE DUPLICATES
df = df.drop_duplicates()
```

```python
# STEP 6 - SELECT VARIABLES
# Quantitative Variables
quantitative_vars = [
    'age',
    'avg_glucose_level',
    'bmi'
]

# Qualitative Variables
qualitative_vars = [
    'gender',
    'hypertension',
    'heart_disease',
    'ever_married',
    'work_type',
    'Residence_type',
    'smoking_status'
]

# Target Variable
target_var = 'stroke'
```

```python
# STEP 7 - ENCODE CATEGORICAL VARIABLES

# Create label encoder
le = LabelEncoder()

# Encode categorical variables
for col in qualitative_vars:
    df[col] = le.fit_transform(df[col])
```

```python
# STEP 8 - FEATURE SCALING

# Create scaler
scaler = StandardScaler()

# Scale numerical variables
df[quantitative_vars] = scaler.fit_transform(df[quantitative_vars])
```

```python
# STEP 9 - SUMMARY STATISTICS

print("\nSummary Statistics")
print(df.describe())
```

```python
# UNIVARIATE ANALYSIS

# STEP 10 - HISTOGRAMS FOR QUANTITATIVE VARIABLES

hist_colors = [
    blue_color,
    green_color,
    red_color
]

for i, var in enumerate(quantitative_vars):

    plt.figure(figsize=(7,5))

    sns.histplot(
        data=df,
        x=var,
        bins=20,
        color=hist_colors[i],
        edgecolor='black'
    )

    plt.title(
        f'Histogram of {var}',
        fontsize=15,
        fontweight='bold'
    )

    plt.xlabel(var, fontsize=12)
    plt.ylabel('Count', fontsize=12)

    plt.show()
```

```python
# STEP 11 - BOXPLOTS FOR QUANTITATIVE VARIABLES

box_colors = [
    blue_color,
    green_color,
    red_color
]

for i, var in enumerate(quantitative_vars):

    plt.figure(figsize=(7,4))

    sns.boxplot(
        x=df[var],
        color=box_colors[i]
    )

    plt.title(
        f'Boxplot of {var}',
        fontsize=15,
        fontweight='bold'
    )

    plt.xlabel(var)

    plt.show()
```

```python
# STEP 12 - BAR CHARTS FOR QUALITATIVE VARIABLES

for var in qualitative_vars:

    plt.figure(figsize=(7,5))

    sns.countplot(
        x=df[var],
        color=blue_color
    )

    plt.title(
        f'Bar Chart of {var}',
        fontsize=15,
        fontweight='bold'
    )

    plt.xlabel(var)
    plt.ylabel('Count')

    plt.xticks(rotation=45)

    plt.show()
```

```python
# STEP 13 - RESPONSE VARIABLE DISTRIBUTION

plt.figure(figsize=(6,4))

sns.countplot(
    x=df[target_var],
    palette=[blue_color, red_color]
)

plt.title(
    'Distribution of Stroke Variable',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Stroke')
plt.ylabel('Count')

plt.show()
```

```python
# BIVARIATE ANALYSIS

# STEP 14 - QUANTITATIVE VARIABLES VS STROKE

for var in quantitative_vars:

    plt.figure(figsize=(7,5))

    sns.boxplot(
        x=df[target_var],
        y=df[var],
        palette=[blue_color, red_color]
    )

    plt.title(
        f'{var} vs Stroke',
        fontsize=15,
        fontweight='bold'
    )

    plt.xlabel('Stroke')
    plt.ylabel(var)

    plt.show()
```

```python
# STEP 15 - QUALITATIVE VARIABLES VS STROKE

for var in qualitative_vars:

    plt.figure(figsize=(8,5))

    sns.countplot(
        x=df[var],
        hue=df[target_var],
        palette=[blue_color, red_color]
    )

    plt.title(
        f'{var} vs Stroke',
        fontsize=15,
        fontweight='bold'
    )

    plt.xlabel(var)
    plt.ylabel('Count')

    plt.xticks(rotation=45)

    plt.legend(title='Stroke')

    plt.show()
```

```python
# STEP 16 - CORRELATION HEATMAP

# Variables for correlation
corr_vars = [
    'age',
    'avg_glucose_level',
    'bmi',
    'hypertension',
    'heart_disease',
    'stroke'
]

# Correlation matrix
corr_matrix = df[corr_vars].corr()

# Plot heatmap
plt.figure(figsize=(8,6))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='Blues',
    fmt='.2f',
    linewidths=0.5
)

plt.title(
    'Correlation Heatmap',
    fontsize=15,
    fontweight='bold'
)

plt.show()
```

```python
# STEP 17 - PAIRPLOT

sns.pairplot(
    df[['age',
        'avg_glucose_level',
        'bmi',
        'stroke']],
    hue='stroke',
    palette='Set1'
)

plt.show()
```

```python
# STEP 18 - CHECK CLASS IMBALANCE

print("\nStroke Value Counts")
print(df['stroke'].value_counts())

print("\nStroke Percentage")
print(df['stroke'].value_counts(normalize=True) * 100)
```

```python
# STEP 18 - VISUALIZE CLASS IMBALANCE


# Count plot for stroke variable

plt.figure(figsize=(6,4))

sns.countplot(
    x=df['stroke'],
    palette=['#1F77B4', '#C0392B']
)

plt.title(
    'Class Distribution of Stroke Variable',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Stroke')
plt.ylabel('Count')

# Add count labels on bars
for p in plt.gca().patches:
    
    plt.gca().annotate(
        f'{int(p.get_height())}',
        (p.get_x() + p.get_width()/2, p.get_height()),
        ha='center',
        va='bottom',
        fontsize=11
    )

plt.show()


# PIE CHART FOR CLASS DISTRIBUTION

stroke_counts = df['stroke'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    stroke_counts,
    labels=['No Stroke', 'Stroke'],
    autopct='%1.1f%%',
    colors=['#1F77B4', '#C0392B'],
    startangle=90
)

plt.title(
    'Percentage Distribution of Stroke Cases',
    fontsize=15,
    fontweight='bold'
)

plt.show()
```
